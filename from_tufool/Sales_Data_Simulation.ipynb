```python
import numpy as np
```

### Sales Data Generation

```python
np.random.seed(42)
```

```python
sales_revenue = 1000 + np.random.normal(loc=0, scale=200, size=30)
sales_revenue = np.clip(sales_revenue, 0, None)  # to ensure no negative sales
```

```python
units_sold = (sales_revenue / 10) + np.random.normal(loc=0, scale=5, size=30)
units_sold = np.clip(units_sold, 0, None)  
units_sold = np.round(units_sold) 
```

### Combine Data

```python
# 2D array: column 1 = revenue, column 2 = units
sales_data = np.column_stack((sales_revenue, units_sold))
```

### KPIs

```python
total_revenue = np.sum(sales_revenue) #month
```

```python
avg_units_per_day = np.mean(units_sold)
```

```python
max_revenue = np.max(sales_revenue)
max_revenue_day = np.argmax(sales_revenue) 
```

```python
total_units = np.sum(units_sold)
avg_revenue_per_unit = total_revenue / total_units if total_units > 0 else 0
```

```python
# display the KPIs
print(f"Total Revenue: ${total_revenue:.2f}")
print(f"Average Units Sold/Day: {avg_units_per_day:.2f}")
print(f"Max Daily Revenue: ${max_revenue:.2f} on Day {max_revenue_day + 1}")
print(f"Average Revenue per Unit: ${avg_revenue_per_unit:.2f}")
```

### Conditional Analysis

```python
# revenue >= 1200
target_revenue = 1200
days_above_target = np.sum(sales_revenue > target_revenue)
```

```python
low_revenue_days_units = units_sold[sales_revenue < 900]
avg_units_low_revenue = np.mean(low_revenue_days_units) if len(low_revenue_days_units) > 0 else 0
```

```python
print(f"Days with Revenue > ${target_revenue}: {days_above_target}")
print(f"Avg Units Sold on Days with Revenue < $900: {avg_units_low_revenue:.2f}")
```

### Weekly Aggregations

```python
# 4 weeks × 7 days = 28 days
weekly_revenue = []
for week in range(4):
    start = week * 7
    end = start + 7
    total_weekly = np.sum(sales_revenue[start:end])
    weekly_revenue.append(total_weekly)
```

```python
for i, rev in enumerate(weekly_revenue, start=1):
    print(f"Week {i}: ${rev:.2f}")
```
