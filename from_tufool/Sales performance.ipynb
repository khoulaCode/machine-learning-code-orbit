```python
train = pd.read_csv("train.csv")
features = pd.read_csv("features.csv")
stores = pd.read_csv("stores.csv")
```

```python
display(features.head())
display(stores.head())
display(train.head())
```

```python
print("\n--- train.describe() ---")
display(train.describe())

print("\n--- train['IsHoliday'].value_counts() ---")
print(train['IsHoliday'].value_counts())

print("\n--- store types value_counts ---")
print(stores["Type"].value_counts())
```

```python
display(train.filter(items=["Store", "Date", "Weekly_Sales"]).head())
display(stores.filter(items=["Store", "Type"]).head())
display(features.filter(items=["Store", "Date", "Temperature"]).head())
```

```python
total_store = train.groupby("Store")["Weekly_Sales"].sum()
print("\n--- Total sales by store ---")
display(total_store)

total_dept = train.groupby("Dept")["Weekly_Sales"].sum()
print("\n--- Total sales by department ---")
display(total_dept)
```

```python
merged_data = pd.merge(train, features, on=["Store", "Date"], how="left")
full_data = pd.merge(merged_data, stores, on="Store", how="left")
display(full_data.head())
```

```python
first_week = train[train['Date'] == train['Date'].min()]
second_week = train[train['Date'] == sorted(train['Date'].unique())[1]]
concatenated_data = pd.concat([first_week, second_week], ignore_index=True)
print("\n--- Concatenated First Two Weeks ---")
display(concatenated_data.head())
```

```python
holiday_sales = train[train['IsHoliday'] == True]
print("\n--- Sales During Holidays ---")
display(holiday_sales[["Store", "Dept", "Date", "Weekly_Sales"]].head())
```
