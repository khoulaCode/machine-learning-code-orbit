```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Time in minutes (0 to 1439)
time_in_minutes = np.arange(0, 24 * 60)
```

```python
np.random.seed(42) 
```

```python
# base 22°C with fluctuations
temperature = 22 + np.random.normal(loc=0, scale=1.5, size=1440)
```

```python
# base 55% with fluctuations, clipped between 0 and 100
humidity = 55 + np.random.normal(loc=0, scale=5, size=1440)
humidity = np.clip(humidity, 0, 100)
```

```python
# base 1012 hPa with fluctuations
pressure = 1012 + np.random.normal(loc=0, scale=2, size=1440)
```

```python
# 2D array [time, temperature, humidity, pressure]
data = np.column_stack((time_in_minutes, temperature, humidity, pressure))
```

```python
# Basic Statistics (24 hours) 
print(f"Temperature: Avg={np.mean(temperature):.2f}°C, Min={np.min(temperature):.2f}°C, Max={np.max(temperature):.2f}°C, Std={np.std(temperature):.2f}")
print(f"Humidity:    Avg={np.mean(humidity):.2f}%, Min={np.min(humidity):.2f}%, Max={np.max(humidity):.2f}%, Std={np.std(humidity):.2f}")
print(f"Pressure:    Avg={np.mean(pressure):.2f}hPa, Min={np.min(pressure):.2f}hPa, Max={np.max(pressure):.2f}hPa, Std={np.std(pressure):.2f}")
```

```python
# 60 minutes per hour
temperature_hourly = temperature.reshape(24, 60)
humidity_hourly = humidity.reshape(24, 60)
pressure_hourly = pressure.reshape(24, 60)
```

```python
avg_temp_per_hour = temperature_hourly.mean(axis=1)
avg_humidity_per_hour = humidity_hourly.mean(axis=1)
avg_pressure_per_hour = pressure_hourly.mean(axis=1)
```

```python
for hour in range(24):
    print(f"Hour {hour:02d}: Temp={avg_temp_per_hour[hour]:.2f}°C, Humidity={avg_humidity_per_hour[hour]:.2f}%, Pressure={avg_pressure_per_hour[hour]:.2f}hPa")
```

```python
# minutes temp > 25°C
anomaly_threshold = 25
anomaly_count = np.sum(temperature > anomaly_threshold)
min_temp_index = np.argmin(temperature)

print(f"Minutes with temperature > {anomaly_threshold}°C: {anomaly_count}")
```

```python
# minute of minimum temperature
min_temp_index = np.argmin(temperature)
print(f"Minute when minimum temperature occurred: {min_temp_index} (Temperature = {temperature[min_temp_index]:.2f}°C)")
```

```python
df = pd.DataFrame(data, columns=["Time (min)", "Temperature (°C)", "Humidity (%)", "Pressure (hPa)"])
df.to_csv("smart_building_data.csv", index=False)
```
