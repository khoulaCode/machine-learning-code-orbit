```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('household_power_consumption.txt', sep =';', na_values ='?')
df
```

```python
df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H:%M:%S')
df.set_index('Datetime', inplace=True)
```

```python
jan_data = df.loc['2008-01']
print(jan_data.head())
```

```python
daily_avg = df['Global_active_power'].resample('D').mean()
weekly_avg = df['Global_active_power'].resample('W').mean()
monthly_avg = df['Global_active_power'].resample('ME').mean()

# Daily plot
plt.figure(figsize=(14, 5))
daily_avg.plot(title='Daily Average Power Consumption')
plt.ylabel('Global Active Power (kilowatts)')
plt.show()

# weekly plot
weekly_avg.plot(title='Weekly Average Power Consumption', figsize=(14, 5))
plt.ylabel('Global Active Power (kilowatts)')
plt.show()

# monthly plot
monthly_avg.plot(title='Monthly Average Power Consumption', figsize=(14,5))
plt.ylabel('Global Active Power (kilowatts)')
plt.show()
```
