```python
import pandas as pd
```

```python
train_df = pd.read_csv("train.csv")
features_df = pd.read_csv("features.csv")
stores_df = pd.read_csv("stores.csv")
```

```python
# EDA
print(train_df.head())
print(features_df.head())
print(stores_df.head())

print(train_df.info())
print(train_df.describe())
print(train_df.isnull().sum())

print(train_df['Store'].value_counts())
print(train_df['Dept'].value_counts())
```

```python
# Merge Datasets
merged_df = pd.merge(train_df, features_df, on=["Store", "Date"], how="left")

merged_df = pd.merge(merged_df, stores_df, on="Store", how="left")

merged_df["Date"] = pd.to_datetime(merged_df["Date"])
```

```python
# GroupBy
# Total
sales_by_store = merged_df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
print(sales_by_store)
# Average 
avg_sales_by_dept = merged_df.groupby("Dept")["Weekly_Sales"].mean()
print(avg_sales_by_dept)
# Monthly 
merged_df["Month"] = merged_df["Date"].dt.month
monthly_sales = merged_df.groupby("Month")["Weekly_Sales"].sum()
print(monthly_sales)

sales_by_type = merged_df.groupby("Type")["Weekly_Sales"].sum()
print(sales_by_type)
```

```python
# Concatenation
week_1 = merged_df[merged_df["Date"] == "2010-02-05"]
week_2 = merged_df[merged_df["Date"] == "2010-02-12"]

combined_weeks = pd.concat([week_1, week_2])
print(combined_weeks.shape)
```

```python
avg_sales = merged_df["Weekly_Sales"].mean()
top_stores = merged_df[merged_df["Weekly_Sales"] > avg_sales]
print(top_stores[["Store", "Dept", "Weekly_Sales"]].head())
```
