```python
# --- Fix strange categorical values ---
df['transaction'] = df['transaction'].where(df['transaction'].isin(['New Property', 'Resale']), np.nan)
df['status'] = df['status'].where(df['status'].str.contains('Ready|Poss.', na=False), np.nan)
df['floor'] = df['floor'].where(df['floor'].astype(str).str.contains('out of', na=False), np.nan)
df['furnishing'] = df['furnishing'].where(df['furnishing'].isin(['Unfurnished', 'Semi-Furnished', 'Furnished']), np.nan)
df['facing'] = df['facing'].where(df['facing'].isin(['East', 'West', 'North - East', 'South - East', 'South -West', 'Main Road', 'Garden/Park']), np.nan)

# --- Fill missing categorical with 'Unknown' ---
df.fillna({'transaction': 'Unknown',
           'status': 'Unknown',
           'floor': 'Unknown',
           'furnishing': 'Unknown',
           'facing': 'Unknown'}, inplace=True)

# --- Filter extreme numeric outliers only ---
df = df[(df['square_feet(sqft)'] <= 5000)]
df = df[(df['price_per_sqft'] >= 1000) & (df['price_per_sqft'] <= 15000)]

# Check how many rows remain
print("✅ Final dataset shape:", df.shape)
```
