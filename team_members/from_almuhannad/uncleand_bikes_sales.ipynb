# Data cleaning

```python
df.drop(columns = ['Day', 'Month', 'Year'],inplace=True)
df
```

```python
df.columns
```

```python
df[df['Order_Quantity'].isnull()]
```

```python
# unit profit = unit_price - unit cost 
# profit = unit_profit * quantity
# quantity = profit / (unit_price - unit_cost)
```

```python
missing = df[df['Order_Quantity'].isnull()]
Profit = missing.Profit
unit_profit = missing.Unit_Price - missing.Unit_Cost
quantity = Profit // unit_profit
quantity
```

```python
df.loc[22,'Order_Quantity'] = quantity.values[0]
df.loc[21:24]
```

```python
missing = df[df['Unit_Cost'] == 0] #to check which column is zero
missing
```

```python
unit_profit = missing.Profit / missing.Order_Quantity
unit_cost = missing.Unit_Price - unit_profit
unit_cost = unit_cost.values
```

```python
df.loc[4, 'Unit_Cost'] = unit_cost[0]
df
```

```python
missing = df[df['Unit_Price'] == 0]
missing
```

```python
unit_price = ((missing.Profit / missing.Order_Quantity) + missing.Unit_Cost).values
df.loc[8, 'Unit_Price'] = unit_price[0]
df.loc[7:9]
```

```python
df.describe()
```

```python
df['Cost'] = df['Unit_Cost'] * df['Order_Quantity']
df['Revenue'] = df['Unit_Price'] * df['Order_Quantity']
df.describe()
```

```python
def set_age_group(age):
    if age<25:
        return 'Youth (<25)'
    elif age<35 :
        return 'Young Adults (25-34)'
    return 'Adults (35-64)'
```

```python
df['Age_Group'] = df['Customer_Age'].apply(set_age_group)
```
