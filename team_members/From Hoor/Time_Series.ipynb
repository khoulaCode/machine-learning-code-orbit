## Resampling (resample('M'), resample('W'))

```python
df['Global_active_power'] = pd.to_numeric(df['Global_active_power'], errors='coerce')

# resample safely
daily_avg = df['Global_active_power'].resample('D').mean()
weekly_sum = df['Global_active_power'].resample('W').sum()
monthly_avg = df['Global_active_power'].resample('M').median()

# Display results
print("Daily Average:\n", daily_avg.head())
print("Weekly Sum:\n", weekly_sum.head())
print("Monthly Median:\n", monthly_avg.head())
```

## Plotting simple trends

```python
import matplotlib.pyplot as plt

daily_avg.plot(title='Daily Average Power Usage')
plt.ylabel('Global Active Power (kilowatts)')
plt.xlabel('Date')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Convert data types
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')  # Convert to datetime
df['Global_active_power'] = pd.to_numeric(df['Global_active_power'], errors='coerce')
df['Global_reactive_power'] = pd.to_numeric(df['Global_reactive_power'], errors='coerce')

# Step 2: Drop rows with missing critical values
df.dropna(subset=['Date', 'Global_active_power', 'Global_reactive_power'], inplace=True)

# Step 3: Sort by Date
df.sort_values('Date', inplace=True)

# Step 4: Plotting
plt.figure(figsize=(14, 6))
plt.plot(df['Date'], df['Global_active_power'], label='Global Active Power (kW)', color='blue', linewidth=0.5)
plt.plot(df['Date'], df['Global_reactive_power'], label='Global Reactive Power (kW)', color='red', linewidth=0.5)
plt.xlabel('Date')
plt.ylabel('Power (kW)')
plt.title('Global Active and Reactive Power Over Time')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
