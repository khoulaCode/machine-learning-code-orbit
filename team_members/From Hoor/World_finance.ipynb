```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
path=r'C:\Users\admin\Desktop\Machine Learning\machine-learning-code-orbit\team_members\From Hoor\Data\Global Economy Indicators.csv'
df = pd.read_csv(path)
df
```

```python
df.columns
```

## 1. Top 5 Countries by Average GDP ---

```python


top5_gdp = df.groupby(' Country ')[' Gross Domestic Product (GDP) '].mean().sort_values(ascending=False).head(5)
print("Top 5 countries by average GDP:")
top5_gdp
```

# --- 2. Top 5 Countries by Household Consumption Expenditure in Most Recent Year ---

```python
# --- 2. Top 5 Countries by Household Consumption Expenditure in Most Recent Year ---

# Assign household consumption column after verifying name above
household_col = ' Household consumption expenditure (including Non-profit institutions serving households) '  # Replace this if needed after checking print(df.columns)

latest_year = df[' Year '].max()
latest_data = df[df[' Year '] == latest_year]

top5_household = latest_data.groupby(' Country ')[household_col].sum().sort_values(ascending=False).head(5)
print(f"\nTop 5 Countries by Household Consumption Expenditure in {latest_year}:")
print(top5_household)
```


# --- 3a. Mean GDP per Capita by Country ---

```python
gdp_per_capita = df.groupby(' Country ')[[" Gross Domestic Product (GDP) "," Per capita GNI "]].mean()
gdp_per_capita
```



# --- 3b. Max Imports of Goods and Services by Country ---

```python

max_imports = df.groupby(' Country ')[ ' Imports of goods and services '].max().sort_values(ascending=False)
max_imports 
```


# --- 3c. Total Exports Over All Years by Country ---

```python
total_exports = df.groupby(' Country ')[' Imports of goods and services '].sum().sort_values(ascending=False)
total_exports
```

# --- 4. Plotting results --

```python
top5_gdp.plot(kind='bar', color='yellow', title='Top 5 Countries by Average GDP')
plt.ylabel('Average GDP')
plt.tight_layout()
plt.show()
```
