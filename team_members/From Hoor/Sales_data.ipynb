```python
import numpy as np

# STEP 1: Generate data
np.random.seed(0)

# Revenue = $1000 ± random between -300 and +300
sales_revenue = 1000 + np.random.randint(-300, 301, size=30)
sales_revenue = np.clip(sales_revenue, 0, None)  # Remove negatives

# Units sold = 50 ± random between -10 and +10
units_sold = 50 + np.random.randint(-10, 11, size=30)
units_sold = np.clip(units_sold, 0, None)  # Remove negatives

# STEP 2: Combine into 2D array
sales_data = np.column_stack((sales_revenue, units_sold))

# STEP 3: Print first 5 days
print("Day | Revenue  | Units Sold")
print("----------------------------")
for i in range(5):
    print(f"{i+1:3d} | ${sales_data[i][0]:8.2f} | {sales_data[i][1]:10d}")
```

```python
# Total revenue
total_revenue = sales_revenue.sum()

# Average units sold
average_units = units_sold.mean()

# Max revenue and day it occurred
max_rev = sales_revenue.max()
day_max_rev = np.argmax(sales_revenue) + 1

# Average revenue per unit
total_units = units_sold.sum()
avg_rev_per_unit = total_revenue / total_units

print("\n--- Key Performance Indicators ---")
print(f"Total Revenue: ${total_revenue}")
print(f"Average Units/Day: {average_units:.2f}")
print(f"Max Revenue: ${max_rev} on Day {day_max_rev}")
print(f"Average Revenue per Unit: ${avg_rev_per_unit:.2f}")
```

```python
# Days with revenue > 1200
high_days = np.sum(sales_revenue > 1200)

# Avg units sold when revenue < 900
low_revenue_mask = sales_revenue < 900
avg_units_low = units_sold[low_revenue_mask].mean()

print("\n--- Conditional Analysis ---")
print(f"Days > $1200 revenue: {high_days}")
print(f"Avg units on low-revenue days (<$900): {avg_units_low:.2f}")
```

```python
# Only include first 28 days = 4 full weeks
weeks = sales_revenue[:28].reshape(4, 7)
weekly_totals = weeks.sum(axis=1)

print("\n--- Weekly Revenue Totals ---")
for i, total in enumerate(weekly_totals, start=1):
    print(f"Week {i}: ${total}")
```
