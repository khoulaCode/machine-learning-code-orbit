```python
train = pd.read_csv(r"C:\Users\admin\Desktop\Machine Learning\machine-learning-code-orbit\team_members\From Hoor\Data\wallmart\train.csv")
train
```

##1. Total sales by store

```python
total_sales_by_store = train.groupby('Store')['Weekly_Sales'].sum()
print(total_sales_by_store.sort_values(ascending=False).head())
```

##2. Total sales by department

```python
total_sales_by_dept = train.groupby('Dept')['Weekly_Sales'].sum()
print(total_sales_by_dept.sort_values(ascending=False).head())
```

##3. Average weekly sales by store

```python
avg_sales_by_store = train.groupby('Store')['Weekly_Sales'].mean()
print(avg_sales_by_store.sort_values(ascending=False).head())
```

##4. Total sales by store and department

```python
total_sales_store_dept = train.groupby(['Store', 'Dept'])['Weekly_Sales'].sum().reset_index()
print(total_sales_store_dept.head())
```

##5. Count of weeks recorded per store

```python
weeks_count_per_store = train.groupby('Store')['Date'].count()
print(weeks_count_per_store)
```

```python
top_stores = total_sales_by_store.sort_values(ascending=False).head(10)
print(top_stores)
```

## 1. Merge sales (train) with store features (features)

```python

# Merge with stores data on 'Store'
final_data = pd.merge(feature, stores, on='Store', how='left')

print(final_data.head())
```

## Concatenating data

```python
first_week = train[train['Date'] == train['Date'].min()]  # Get the first week of data
second_week = train[train['Date'] == train['Date'].sort_values().unique()[1]]  # Get the second week of data
concatenated_data = pd.concat([first_week, second_week], ignore_index=True) # Concatenate the two weeks of data
concatenated_data.head()
```
