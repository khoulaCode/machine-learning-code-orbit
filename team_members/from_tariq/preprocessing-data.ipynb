```python
fig, (inches, price) = plt.subplots(1, 2, figsize=(12, 5))
inches.boxplot(df['Inches'])
inches.set_title('Boxplot of Inches')
inches.set_ylabel('Inches')
price.boxplot(df['Price'])
price.set_title('Boxplot of Price')
price.set_ylabel('Price')
plt.tight_layout()
plt.show()
```

```python
df['Inches_Std'] = (df['Inches'] - df['Inches'].mean()) / df['Inches'].std()
df['Price_Std'] = (df['Price'] - df['Price'].mean()) / df['Price'].std()
df['Inches_Std']
df['Price_Std']
```

```python

df['Inches_Norm'] = (df['Inches'] - df['Inches'].min()) / (df['Inches'].max() - df['Inches'].min())
df['Price_Norm'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())
df['Inches_Norm']
df['Price_Norm']
```

```python
df['Ram'] = df['Ram'].str.replace('GB', '').astype(int)
df['Weight'] = df['Weight'].str.replace('kg', '').astype(float)
df
```
