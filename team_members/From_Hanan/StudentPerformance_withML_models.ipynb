# Student Performance Dataset Analysis Report

## 1. Introduction

This report analyzes a student performance dataset consisting of features that potentially influence academic success. The goal is to predict the **Performance Index** based on various factors such as study habits, prior scores, extracurricular activities, and lifestyle factors.

---

## 2. Dataset Description

The dataset contains the following columns:

| Column Name                      | Data Type | Description                              |
|---------------------------------|-----------|------------------------------------------|
| Hours Studied                   | int64     | Number of hours the student studied     |
| Previous Scores                 | int64     | Scores from previous assessments         |
| Extracurricular Activities      | object    | Participation in extracurricular activities (Yes/No) |
| Sleep Hours                    | int64     | Average number of hours of sleep          |
| Sample Question Papers Practiced | int64     | Number of sample question papers practiced |
| Performance Index               | float64   | Target variable representing overall performance |

---

## 3. Data Preprocessing

- The categorical variable **Extracurricular Activities** was encoded using Label Encoding (Yes = 1, No = 0).
- Feature scaling was applied using **Min-Max Scaling** to normalize all feature values between 0 and 1.
- The target variable **Performance Index** was kept unscaled.
- The dataset was split into training and testing sets with a 70:30 ratio using a random state of 42 for reproducibility.

---

## 4. Regression Models and Evaluation

The following regression models were trained and evaluated to predict the student **Performance Index**:

- Linear Regression  
- Polynomial Regression (Degree 2)  
- Lasso Regression  
- Ridge Regression  
- Decision Tree Regressor  
- Random Forest Regressor  

The performance metrics used include:
- Mean Absolute Error (MAE)  
- Mean Squared Error (MSE)  
- Root Mean Squared Error (RMSE)  
- R-squared (R²)

---

## 5. Results (R-squared Scores)

| Model                          | R² Score |
|--------------------------------|----------|
| Linear Regression              | 0.99     |
| Polynomial Regression (Degree 2) | 0.99     |
| Lasso Regression               | 0.99     |
| Ridge Regression              | 0.99     |
| Decision Tree Regressor         | 0.98     |
| Random Forest Regressor         | 0.99     |

---

## 6. Conclusion

All models showed strong predictive performance on the dataset, with R² values close to 1.0 indicating a very good fit. The ensemble model (Random Forest) and regularized regressions (Lasso, Ridge) performed comparably to basic linear models. Decision Trees showed slightly lower performance, which might be improved with hyperparameter tuning.

# The Code:

```python
#import Required Library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, LabelEncoder, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
#Read file and show head
df = pd.read_csv("Student_Performance.csv")
df.head()
```

```python
#Eplore the Datatypes
df.dtypes
```

```python
#Check Null Values
df.isnull().sum()
```

```python
#Encode categorical column using label Encodel
encoder = LabelEncoder()
df['Extracurricular Activities'] = encoder.fit_transform(df['Extracurricular Activities'])
```

```python
#Separate target column (y) and data dataframe (X)
y = df['Performance Index']
X = df.drop('Performance Index', axis=1)
```

```python
#Scale the features using MinMaxScaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Helper function to compute and print evaluation metrics for the models models
def evaluate_model(name, model, X_test, y_test, y_pred):
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    print(f"\n--- {name} ---")
    print(f"MAE: {mae:.2f}")
    print(f"MSE: {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"R2: {r2:.2f}")
```

```python
#split into Train and test
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

```python
# Linear Regression Model
lr = LinearRegression()
lr.fit(X_train, y_train)
evaluate_model("Linear Regression", lr, X_test, y_test, lr.predict(X_test))
```

```python
# Polynomial Regression (degree 2) Model
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

poly_lr = LinearRegression()
poly_lr.fit(X_train_poly, y_train)
evaluate_model("Polynomial Regression (Degree 2)", poly_lr, X_test, y_test, poly_lr.predict(X_test_poly))
```

```python
# Lasso Regression Model
lasso = Lasso(alpha=0.1, max_iter=10000)
lasso.fit(X_train, y_train)
evaluate_model("Lasso Regression", lasso, X_test, y_test, lasso.predict(X_test))
```

```python
# Ridge Regression Model
ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)
evaluate_model("Ridge Regression", ridge, X_test, y_test, ridge.predict(X_test))
```

```python
# Decision Tree Model
tree = DecisionTreeRegressor(random_state=42)
tree.fit(X_train, y_train)
evaluate_model("Decision Tree", tree, X_test, y_test, tree.predict(X_test))
```

```python
# Random Forest Model
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
evaluate_model("Random Forest", rf, X_test, y_test, rf.predict(X_test))
```
