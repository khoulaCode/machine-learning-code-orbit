```python
import numpy as np
```

```python
sales_data = np.random.normal(90, 1000, size=30)
sales_data = np.round(sales_data, 2)
sales_data = np.clip(sales_data, 0, None)
print("Sales data: ", sales_data)
```

```python
units = np.random.randint(10, 100, size=30)
print("Units sold: ", units)
```

```python
revenue_sales = sales_data * units
print("Sales revenue: ", revenue_sales)
```

```python
combine_data_2d = np.column_stack((revenue_sales, units))
print("comine data:",combine_data_2d)
```

```python
cal_total_sales = np.sum(revenue_sales) 
print("Total sales revenue for the month: ", cal_total_sales)
```

```python
cal_average_sales = np.mean(units)
cal_average_sales = np.round(cal_average_sales, 0)
cal_average_sales = int(cal_average_sales)
print("Average sales units_sold for the month:", cal_average_sales)
```

```python
cal_max_sales = np.max(revenue_sales)
cal_max_sales_index = np.argmax(revenue_sales)
print(f"Maximum sales revenue of {cal_max_sales:.2f} occurred at index {cal_max_sales_index}.")
```

```python
total_units_sold = np.sum(units)
average_revenue_per_unit = cal_total_sales / total_units_sold
print(f"Average revenue for the month: {average_revenue_per_unit:.2f}")
```

```python
sales_above_target_per_days = np.sum(revenue_sales > 1200)
print("Number of days sales revenue more than 1200: ",sales_above_target_per_days)
```

```python
units_sold_when_revenue_blow_target = np.sum(units[revenue_sales < 900])
print(f"Units sold when sales revenue was below 900: {units_sold_when_revenue_blow_target}")
```

```python
weekly_sales_rev = np.array([])
for i in range(4):
    start = i * 7
    end = start + 7
    week_total = np.sum(revenue_sales[start:end])
    weekly_sales_rev = np.append(weekly_sales_rev, week_total)
print("Total sales revenue for full week:", weekly_sales_rev)
```
