```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
path = r"..\..\data\household_power_consumption.txt"
df = pd.read_csv(path, sep=';')
df
```

```python
df.columns
```

```python
df.index
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Time'] = pd.to_datetime(df['Time'])
df
```

```python
df.info()
```

```python
df['Datetime'] = pd.to_datetime(df['Date'], ' ' , df['Time'])
df.set_index('Datetime')
df.head()
```

```python
df_timebased = df.set_index('Datetime').loc['2006-12-16']
df_timebased.head()
```

```python
resampled_month = df.set_index('Datetime').resample('M')
resampled_week = df.set_index('Datetime').resample('W')

print("Monthly resampled :")
print(resampled_month)

print("Weekly resampled :")
print(resampled_week)
```

```python
df['Global_active_power'] = df['Global_active_power'].replace('?', np.nan).astype(float)
df['Global_reactive_power'] = df['Global_reactive_power'].replace('?', np.nan).astype(float)
df['Voltage'] = df['Voltage'].replace('?', np.nan).astype(float)
df['Global_intensity'] = df['Global_intensity'].replace('?', np.nan).astype(float)
df['Sub_metering_1'] = df['Sub_metering_1'].replace('?', np.nan).astype(float)
df['Sub_metering_2'] = df['Sub_metering_2'].replace('?', np.nan).astype(float)

plt.figure(figsize=(10, 5))
plt.plot(df['Datetime'], df['Global_active_power'], label='Global Active Power', color='black')
plt.plot(df['Datetime'], df['Global_reactive_power'], label='Global Reactive Power', color='blue')
plt.xlabel('Datetime')
plt.ylabel('Power (kW)')
plt.title('Global Active and Reactive Power Over Time')
plt.show()
```
