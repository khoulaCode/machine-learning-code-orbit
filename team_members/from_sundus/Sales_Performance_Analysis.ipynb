```python
# Merge train with features 
train_features = pd.merge(train, features, on=['Store', 'Date'])
```

```python
full_data = pd.merge(train_features, stores, on='Store') 
full_data.head() 
```

```python
#Total Sales by Store 
sales_by_store = full_data.groupby('Store')['Weekly_Sales'].sum().reset_index()
sales_by_store = sales_by_store.sort_values(by='Weekly_Sales', ascending=False)
print("Total Sales by Store")
sales_by_store
```

```python
#Total Sales by Department
sales_by_dept = full_data.groupby('Dept')['Weekly_Sales'].sum().reset_index()
sales_by_dept = sales_by_dept.sort_values(by='Weekly_Sales', ascending=False)
print("Total Sales by Department")
sales_by_dept
```

```python
#Average Sales per Store Type
avg_sales = full_data.groupby('Type')['Weekly_Sales'].mean().reset_index()
print("Average Sales by Store Type")
avg_sales
```

```python
week1 = full_data[full_data['Date'] == '2010-02-05']
week2 = full_data[full_data['Date'] == '2010-02-12'] 
combined_weeks = pd.concat([week1, week2])
combined_weeks
```
