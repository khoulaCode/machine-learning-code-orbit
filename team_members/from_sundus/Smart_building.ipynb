```python
import numpy as np 
```

```python
time_in_minutes = np.arange(1440) 
print("Time in minutes:", time_in_minutes[:10]) 
```

```python
temperature = 22 + np.random.normal(0, 1, 1440) 
print("Sample temperature values:", temperature[:10]) 
```

```python
humidity = 55 + np.random.normal(0, 5, 1440)
humidity = np.clip(humidity, 0, 100)
print("Sample humidity values:", humidity[:10]) 
```

```python
pressure = 1012 + np.random.normal(0, 3, 1440)
print("Sample pressure values:", pressure[:10]) 
```

```python
data = np.column_stack((time_in_minutes, temperature, humidity, pressure))
print("First 5 rows of data:\n", data[:5]) 
```

```python
print("Average temperature:", np.mean(temperature))
print("Minimum temperature:", np.min(temperature))
print("Maximum temperature:", np.max(temperature)) 
```

```python
print("Average humidity:", np.mean(humidity))
print("Minimum humidity:", np.min(humidity))
print("Maximum humidity:", np.max(humidity))
```

```python
print("Average pressure:", np.mean(pressure))
print("Minimum pressure:", np.min(pressure))
print("Maximum pressure:", np.max(pressure)) 
```

```python
print("Standard deviation of temperature:", np.std(temperature)) 
```

```python
print("Standard deviation of humidity:", np.std(humidity)) 
```

```python
print("Standard deviation of pressure:", np.std(pressure))
```

```python
temperature_hourly = temperature.reshape(24, 60).mean(axis=1) 
print("Hourly average temperature:\n", temperature_hourly) 
```

```python
humidity_hourly = humidity.reshape(24, 60).mean(axis=1) 
print("\nHourly average humidity:\n", humidity_hourly) 
```

```python
pressure_hourly = pressure.reshape(24, 60).mean(axis=1)
print("\nHourly average pressure:\n", pressure_hourly)
```

```python
high_temp_count = np.sum(temperature > 25)
print("Number of minutes where temperature > 25°C:", high_temp_count) 
```

```python
min_temp_time = time_in_minutes[np.argmin(temperature)]
print("Minimum temperature occurred at minute:", min_temp_time) 
```

```python
np.savetxt("environment_data.csv", data, delimiter=",", header="Time,Temperature,Humidity,Pressure", comments='', fmt="%.2f")
print("Data saved to 'environment_data.csv'")
```
