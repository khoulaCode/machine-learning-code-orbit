```python
import numpy as np
```

```python
np.random.seed(0)
```

```python
sales_revenue = np.random.normal(loc=1000, scale=200, size=30) 
sales_revenue = np.clip(sales_revenue, 0, None)
sales_revenue
```

```python
sales_revenue = np.round(sales_revenue, 2)
print("Sales Revenue for 30 Days:\n", sales_revenue)
```

```python
# Create units sold (roughly sales_revenue / 10)
units_sold = (sales_revenue / 10) + np.random.normal(loc=0, scale=5, size=30)
units_sold = np.clip(units_sold, 0, None)
units_sold = np.round(units_sold, 2)
print("Units Sold for 30 Days:\n", units_sold)
```

```python
#Combine Data in One Array
sales_data = np.column_stack((sales_revenue, units_sold))
sales_data
```

```python
#Calculate Key Performance Indicators
total_revenue = np.sum(sales_revenue)
print("Total Revenue for 30 Days: ", round(total_revenue, 2))
```

```python
average_units = np.mean(units_sold)
print("Average Units Sold per Day:", round(average_units, 2))
```

```python
max_revenue = np.max(sales_revenue)
max_day = np.argmax(sales_revenue) + 1 # day start by number 1 
print("Maximum Revenue: ", max_revenue, "on Day", max_day)
```

```python
total_units = np.sum(units_sold)
avg_revenue_per_unit = total_revenue / total_units 
print("Average Revenue per Unit Sold:", round(avg_revenue_per_unit, 2))
```

```python
# when revenue is above $1200
high_revenue_days = sales_revenue > 1200
count_high_days = np.sum(high_revenue_days)
print("Number of Days with Revenue > 1200:", count_high_days)
```

```python
# when revenue is below $900
low_revenue_days = sales_revenue < 900
avg_units_low_days = np.mean(units_sold[low_revenue_days])
print("Average Units Sold when revenue below 900:", round(avg_units_low_days, 2)) 
```

```python
#Weekly Revenue Totals 
weekly_revenue = sales_revenue[:28].reshape(4, 7) 
weekly_totals = np.sum(weekly_revenue, axis=1)
for i, total in enumerate(weekly_totals, start=1):
    print(f"Week {i} Total Revenue: ${round(total, 2)}") 
```
