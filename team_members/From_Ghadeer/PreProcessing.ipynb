```python
import pandas as pd
import numpy as np
```

```python
# Load your CSV file
path = r"C:\Users\admin\Desktop\archive (1)\laptopData.csv"
df= pd.read_csv(path)
df 
```

```python
# Step 1: Drop any unneeded index column (like 'Unnamed: 0')
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

df
```

```python
# Step 2: Handle missing values and invalid entries
df = df.replace('?', np.nan)  # Replace '?' with NaN
df = df.dropna()              # Drop rows with missing data
df
```

```python
# Step 3: Convert 'Ram' column from "8GB" to integer
df['Ram'] = df['Ram'].astype(str).str.replace('GB', '', regex=False).astype(int)
```

```python
# Step 4: Convert 'Weight' from "2.1kg" to float
df['Weight'] = df['Weight'].str.replace('kg', '', regex=False)
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce')  # convert to float
df = df.dropna(subset=['Weight'])  # drop rows with invalid weight
```

```python
# Step 5: Convert 'Inches' to float
df['Inches'] = df['Inches'].astype(float)

# Step 6: Extract CPU brand (first word of the CPU string)
df['Cpu_Brand'] = df['Cpu'].apply(lambda x: x.split()[0])
```

```python
# Step 7: Clean the 'Memory' column to split into SSD and HDD

# Normalize memory labels
df['Memory'] = df['Memory'].str.replace('Flash Storage', 'SSD')
df['Memory'] = df['Memory'].str.replace('Hybrid', 'HDD')
```

```python
def extract_storage(mem_str): 
    ssd = 0
    hdd = 0
    if pd.isna(mem_str):
        return pd.Series([ssd, hdd])

    parts = mem_str.split('+')
    for part in parts:
        part = part.strip()
        if 'SSD' in part:
            value = part.replace('SSD', '').replace('GB', '').replace('TB', '')
            if 'TB' in part:
                ssd += float(value.strip()) * 1000
            else:
                ssd += float(value.strip())
        elif 'HDD' in part:
            value = part.replace('HDD', '').replace('GB', '').replace('TB', '')
            if 'TB' in part:
                hdd += float(value.strip()) * 1000
            else:
                hdd += float(value.strip())
    return pd.Series([int(ssd), int(hdd)])

# Apply it safely to the column
df[['SSD', 'HDD']] = df['Memory'].apply(extract_storage)
df = df.drop(columns=['Memory'])
```

```python
# ✅ Final cleaned data ready
print("✅ Data cleaned successfully!")
print(df.head())
```
