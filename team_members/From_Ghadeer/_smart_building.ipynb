```python
import numpy as np
```

## Data Generation

```python

# 1. Data Generation
time_in_minutes = np.arange(0, 1440)  # 24 hours * 60 minutes

# Generate temperature: base 22°C with some random fluctuation (mean=0, std=1)
temperature = 22 + np.random.normal(0, 1, size=1440)

# Generate humidity: base 55% with some fluctuation (keep in 0-100%)
humidity = 55 + np.random.normal(0, 5, size=1440)
humidity = np.clip(humidity, 0, 100)  # ensure 0% ≤ humidity ≤ 100%

# Generate pressure: base 1012 hPa with some fluctuation
pressure = 1012 + np.random.normal(0, 2, size=1440)

# Combine into 2D array: [time, temperature, humidity, pressure]
data = np.column_stack((time_in_minutes, temperature, humidity, pressure))
```

##  Basic Statistics

```python
# Average values
avg_temp = np.mean(temperature)
avg_hum = np.mean(humidity)
avg_pres = np.mean(pressure)

# Min and max values
min_temp = np.min(temperature)
max_temp = np.max(temperature)
min_hum = np.min(humidity)
max_hum = np.max(humidity)
min_pres = np.min(pressure)
max_pres = np.max(pressure)

# Standard deviation
std_temp = np.std(temperature)
std_hum = np.std(humidity)
std_pres = np.std(pressure)

# Display results
print("Basic Statistics:")
print(f"Temperature - Avg: {avg_temp:.2f}, Min: {min_temp:.2f}, Max: {max_temp:.2f}, Std: {std_temp:.2f}")
print(f"Humidity    - Avg: {avg_hum:.2f}, Min: {min_hum:.2f}, Max: {max_hum:.2f}, Std: {std_hum:.2f}")
print(f"Pressure    - Avg: {avg_pres:.2f}, Min: {min_pres:.2f}, Max: {max_pres:.2f}, Std: {std_pres:.2f}")
```

## Hourly Averages

```python
# Reshape into (24 hours, 60 minutes per hour)
temp_hourly = temperature.reshape(24, 60)
hum_hourly = humidity.reshape(24, 60)
pres_hourly = pressure.reshape(24, 60)

# Calculate hourly averages
avg_temp_hourly = np.mean(temp_hourly, axis=1)
avg_hum_hourly = np.mean(hum_hourly, axis=1)
avg_pres_hourly = np.mean(pres_hourly, axis=1)

# Display first few hourly averages
print("\nHourly Averages (First 5 hours):")
for i in range(5):
    print(f"Hour {i}: Temp={avg_temp_hourly[i]:.2f}, Hum={avg_hum_hourly[i]:.2f}, Pres={avg_pres_hourly[i]:.2f}")
```

## Anomaly Detection

```python
# Count how many minutes temperature > 25°C
high_temp_minutes = np.sum(temperature > 25)

# Find time (minute) of minimum temperature
min_temp_time = time_in_minutes[np.argmin(temperature)]

print("Minutes with temperature > 25°C:", high_temp_minutes)
print("Time (minute) of minimum temperature:", min_temp_time)
##
```
