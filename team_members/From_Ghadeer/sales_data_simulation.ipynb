## Generate Sales Data

```python
import numpy as np
```

```python
np.random.seed(42)  # for reproducible results

# 1. Sales Data Generation
days = 30
base_revenue = 1000

# Generate sales_revenue with random normal fluctuation (mean=0, std=200)
sales_revenue = base_revenue + np.random.normal(0, 200, size=days)
sales_revenue = np.clip(sales_revenue, 0, None)  # no negative sales

# Units sold: correlated with revenue but with its own fluctuation
base_units = 100
units_sold = (sales_revenue / base_revenue) * base_units + np.random.normal(0, 10, size=days)
units_sold = np.clip(units_sold, 0, None).astype(int)  # no negative units, convert to int

# Display first few
print("Sales Revenue (first 5 days):", sales_revenue[:5])
print("Units Sold (first 5 days):", units_sold[:5])
```

## Combine into 2D Array

```python
# Combine into [sales_revenue, units_sold]
sales_data = np.column_stack((sales_revenue, units_sold))
print("\nCombined Sales Data (first 5 rows):\n", sales_data[:5])
```

## Key Performance Indicators (KPIs)

```python
# Total revenue
total_revenue = np.sum(sales_revenue)

# Average units sold per day
avg_units_per_day = np.mean(units_sold)

# Max daily revenue and the day it occurred
max_revenue = np.max(sales_revenue)
max_day = np.argmax(sales_revenue)

# Average revenue per unit
total_units = np.sum(units_sold)
avg_revenue_per_unit = total_revenue / total_units

# Print KPIs
print("\n--- KPIs ---")
print(f"Total Revenue for 30 days: ${total_revenue:.2f}")
print(f"Average Units Sold per Day: {avg_units_per_day:.2f}")
print(f"Max Daily Revenue: ${max_revenue:.2f} (Day {max_day + 1})")
print(f"Average Revenue per Unit: ${avg_revenue_per_unit:.2f}")
```

## Conditional Analysis

```python
# Days with revenue > $1200
high_sales_days = np.sum(sales_revenue > 1200)

# Average units sold when revenue < $900
low_days = sales_revenue < 900
avg_units_low_days = np.mean(units_sold[low_days]) if np.any(low_days) else 0

print("\n--- Conditional Analysis ---")
print(f"Days with Revenue > $1200: {high_sales_days}")
print(f"Average Units Sold when Revenue < $900: {avg_units_low_days:.2f}")
```

## Weekly Aggregation (4 weeks: Mon–Sun)

```python
# Weeks: [0:7], [7:14], [14:21], [21:28]
weekly_revenue = [
    np.sum(sales_revenue[0:7]),
    np.sum(sales_revenue[7:14]),
    np.sum(sales_revenue[14:21]),
    np.sum(sales_revenue[21:28])
]

print("\n--- Weekly Revenue ---")
for i, total in enumerate(weekly_revenue):
    print(f"Week {i + 1} Revenue: ${total:.2f}")
```
