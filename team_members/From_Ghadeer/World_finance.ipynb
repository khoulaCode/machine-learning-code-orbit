```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 1: Load the data
path= r"C:\Users\admin\Desktop\archive (1)\Global Economy Indicators.csv"
df = pd.read_csv(path) # Make sure the CSV file is in your folder

# Show rows
df
```

```python
top5_gdp = df.groupby(' Country ')[' Gross Domestic Product (GDP) '].mean().sort_values(ascending=False).head(5)
print("Top 5 countries by average GDP:\n", top5_gdp)

# Plot
top5_gdp.plot(kind='bar', title="Top 5 Countries by Avg GDP", ylabel="Average GDP", figsize=(8,5))
plt.show()
```

```python
# Find the latest year in dataset
latest_year = df[' Year '].max()

# Filter for that year and drop NA
recent_household = df[df[' Year '] == latest_year][[' Country ', ' Household consumption expenditure (including Non-profit institutions serving households) ']].dropna()

print(" Countries with household expenditure in ", latest_year, ":\n", recent_household)

# Plot top 10
recent_household.sort_values(by=' Household consumption expenditure (including Non-profit institutions serving households) ', ascending=False).head(10).plot(
    kind='bar', x=' Country ', y=' Household consumption expenditure (including Non-profit institutions serving households) ',
    title="Top 10 Countries by Household Expenditure in Latest Year", figsize=(10,5)
)
plt.show()
```

```python
grouped = df.groupby(' Country ').agg({
    ' Gross Domestic Product (GDP) ': "mean",
    ' Imports of goods and services ': "max",
    ' Exports of goods and services ': "sum"
}).reset_index()

# Rename for clarity
grouped.columns = ["Country", "Mean GDP per Capita", "Max Imports", "Total Exports"]
print(grouped.head())
```

```python
# Top 10 countries by Mean GDP per Capita
grouped.sort_values(by="Mean GDP per Capita", ascending=False).head(10).plot(
    kind='bar', x='Country', y='Mean GDP per Capita',
    title="Top 10 Countries by Mean GDP per Capita", figsize=(10,5)
)
plt.show()

# Top 10 by Max Imports
grouped.sort_values(by="Max Imports", ascending=False).head(10).plot(
    kind='bar', x='Country', y='Max Imports',
    title="Top 10 Countries by Max Imports", figsize=(10,5)
)
plt.show()

# Top 10 by Total Exports
grouped.sort_values(by="Total Exports", ascending=False).head(10).plot(
    kind='bar', x='Country', y='Total Exports',
    title="Top 10 Countries by Total Exports", figsize=(10,5)
)
plt.show()
```
