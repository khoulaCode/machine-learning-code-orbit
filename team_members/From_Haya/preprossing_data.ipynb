```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\admin\Desktop\data\laptopData.csv")
df.head()
```

```python
df.describe()  #Check the summary statistics for numeric columns
```

```python
df.isnull().sum() #Check for missing (null) values in each column
```

```python
df.dropna(inplace=True)   #drop rows with missing values
```

```python
# Drop unnecessary column
df.drop(columns=["Unnamed: 0"], inplace=True)

# Remove rows with missing values in key columns
df = df.dropna(subset=["Ram", "Weight", "Cpu", "Gpu", "Memory", "ScreenResolution", "OpSys", "Price"])
df
```

```python
df.duplicated().sum() 
```

```python
# Clean 'Ram' column (remove 'GB' and convert to int)
df["Ram"] = df["Ram"].astype(str).str.replace("GB", "", regex=False).astype(int)

df.head(10)
```

```python
df.info() 
```

```python
# Clean 'Weight' column (remove 'kg' and convert to float, replace '?' with NaN)
df['Weight']= df['Weight'].replace('?', np.nan)
df['Weight'] = df["Weight"].str.replace('kg', '', regex=False).astype(float)


df.info()
```

```python
# Extract screen resolution width and height
df["Resolution_Width"] = df["ScreenResolution"].str.extract(r'(\d+)x\d+').astype(float)
df["Resolution_Height"] = df["ScreenResolution"].str.extract(r'\d+x(\d+)').astype(float)
df.head()
```

```python
# Identify types of storage
df["SSD"] = df["Memory"].str.contains("SSD").astype(int)
df["HDD"] = df["Memory"].str.contains("HDD").astype(int)
df["Flash_Storage"] = df["Memory"].str.contains("Flash Storage").astype(int)
df["Hybrid"] = df["Memory"].str.contains("Hybrid").astype(int)
df.head()
```

```python
print(df["Inches"].unique())

# # Convert Inches column to float
df["Inches"] = df["Inches"].replace("?", np.nan)
df["Inches"] = pd.to_numeric(df["Inches"], errors='coerce')



df.head()
```

```python
print(df["Inches"].unique())
```

```python
print(df["Inches"].dtypes)
```

```python
# Extract CPU brand and speed
df["Cpu_Brand"] = df["Cpu"].apply(lambda x: x.split()[0] + " " + x.split()[1])
df["Cpu_Speed"] = df["Cpu"].str.extract(r'(\d\.\d+)GHz').astype(float)

df.head(10)
```

```python
# GPU Brand
df["Gpu_Brand"] = df["Gpu"].apply(lambda x: x.split()[0])
df.head(10)
```

```python
# Clean and normalize operating system names
df["OpSys"] = df["OpSys"].str.strip().str.lower()
df.head(10)
```

```python
# Convert Price to float
df["Price"] = df["Price"].astype(float)
df.info()
```

```python
# Preview some values in ScreenResolution column
print(df["ScreenResolution"].unique()[:5])

# Extract screen width and height
df["Resolution_Width"] = df["ScreenResolution"].str.extract(r'(\d+)x')[0].astype(float)
df["Resolution_Height"] = df["ScreenResolution"].str.extract(r'x(\d+)')[0].astype(float)

# Identify if the screen is IPS
df["Has_IPS"] = df["ScreenResolution"].str.contains("IPS", case=False, na=False).astype(int)

# Identify if the screen is Touchscreen
df["Has_Touchscreen"] = df["ScreenResolution"].str.contains("Touchscreen", case=False, na=False).astype(int)

# Optional: Calculate Pixels Per Inch (PPI) = screen resolution density
df["PPI"] = ((df["Resolution_Width"]**2 + df["Resolution_Height"]**2)**0.5) / df["Inches"]

# Display some results
df[["ScreenResolution", "Resolution_Width", "Resolution_Height", "Has_IPS", "Has_Touchscreen", "PPI"]].head()
```

```python
# List of original columns that were already processed or split
columns_to_drop = [
    "ScreenResolution",  # split into Resolution_Width, Resolution_Height, Has_IPS, Has_Touchscreen
    "Memory",            # split into SSD, HDD, Flash_Storage, Hybrid
    "Cpu",               # split into Cpu_Brand and Cpu_Speed
    "Gpu"                # split into Gpu_Brand
]
# Drop the redundant columns
df = df.drop(columns=columns_to_drop)
# Display the cleaned dataframe
df.head(10)
```

## Histogram: Price Distribution

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Price"], bins=30)
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
```

##  Scatter Plot: Price vs RAM

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Ram"], df["Price"])
plt.title("Price vs RAM")
plt.xlabel("RAM (GB)")
plt.ylabel("Price")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Boxplot: Price by Operating System

```python
plt.figure(figsize=(10, 6))
df.boxplot(column="Price", by="OpSys", rot=45)
plt.title("Price Distribution by Operating System")
plt.suptitle("")
plt.xlabel("Operating System")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```
