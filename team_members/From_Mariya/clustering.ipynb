```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv(r"C:\Users\admin\OneDrive\Desktop\ML\machine-learning-code-orbit\team_members\From_Mariya\data\Mall_Customers.csv")
print(df.head())
```

```python
X = df[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
inertia = []
k_range = range(1, 11)

for k in k_range:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X)
    inertia.append(model.inertia_)

plt.plot(k_range, inertia, marker='o')
plt.title("Elbow Method - Find Best k")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.grid(True)
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, random_state=42)
df['Cluster'] = kmeans.fit_predict(X)
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x='Annual Income (k$)',
    y='Spending Score (1-100)',
    hue='Cluster',
    data=df,
    palette='Set1',
    s=100
)
plt.title("Customer Groups (K-Means Clustering)")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score")
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```
